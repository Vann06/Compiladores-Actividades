# Ejercicio1 
Sea X una variable aleatoria exponencial con media 1. Proporcione un algoritmo eficiente para
simular una variable aleatoria cuya distribución es la distribución condicional de X dado que X <
0,05. Es decir, su función de densidad es 

Genere 1000 de estas variables y utilícelas para estimar E[X | X < 0,05]. Luego determine el
valor exacto de E[X | X < 0,05]

## Análisis:
Solo nos importan valores dentro del rango 0 < X < 0.05, que cumplen la condición. Debemos de encontrar la función acumulada:
 
 $$F(x) = \frac{1 - e^{-x}}{1 - e^{-0.05}}$$
para $0 < x < 0.05$

Ahora generamos una variable aleatoria uniforme $U$ en el intervalo $(0,1)$ y la transformamos usando la función inversa de $F(x)$:

$$U \sim \text{Uniforme}(0,1)$$

Para encontrar $X = F^{-1}(U)$, despejamos $x$ de la ecuación $U = F(x)$:

$$U = \frac{1 - e^{-x}}{1 - e^{-0.05}}$$

$$U(1 - e^{-0.05}) = 1 - e^{-x}$$

$$e^{-x} = 1 - U(1 - e^{-0.05})$$

$$-x = \ln[1 - U(1 - e^{-0.05})]$$

$$X = -\ln[1 - U(1 - e^{-0.05})]$$

## Explicación del algoritmo: 
1. Generar U uniforme en (0,1)
2. Calcular X usando la fórmula $X = -\ln[1 - U(1 - e^{-0.05})]$
3. Retornar X como la variable aleatoria simulada.

y ahora solo se generan 1000 de estas variables para estimar $E[X | X < 0.05]$ y luego determinar el valor exacto de $E[X | X < 0.05]$.

    

````python
n = 1000
suma = 0

# el valor esperado de X dado que X < 0.05 se puede estimar 
# generando n variables aleatorias uniformes U en el
# intervalo (0,1) y transformándolas usando la función inversa de F(x):
for i in range(n):
    u = random.random()

    x = -math.log(1 - u * (1 - math.exp(-0.05)))

    suma += x

promedio = suma / n

print("Estimación de E[X | X < 0.05]:", promedio)

# valor exacto de E[X | X < 0.05] se puede calcular usando la integral de la función de densidad condicional:
exacto = (1 - 1.05 * math.exp(-0.05)) / (1 - math.exp(-0.05))
print("Valor exacto de E[X | X < 0.05]:", exacto)
````

# Ejercicio 2
Suponga que es relativamente fácil generar variables aleatorias a partir de cualquiera de las dis-
tribuciones Fi, i = 1, . . . , n. ¿Cómo podríamos generar una variable aleatoria con función de distribución

$$F(x) = \sum_{i=1}^{n} p_i(x)$$

donde pi, i = 1, . . . , n, son números no negativos cuya suma es 1?

## Análisis: 

Generamos una distribución uniforme U ~ Uniforme(0,1) y contruimos intervalos acumulados correpondiendo a cada una de las distribuciones Fi, i = 1, . . . , n. Luego, dependiendo del valor de U, seleccionamos la distribución correspondiente y generamos una variable aleatoria a partir de esa distribución.

Si tuvieramos U = 0.37 y con intervalos [0,0.2) para F1, [0.2,0.5) para F2 y [0.5,1) para F3, entonces U = 0.37 cae en el intervalo de F2, por lo que generamos una variable aleatoria a partir de la distribución F2.

un algoritmo para esto sería: 
1. Generar U uniforme en (0,1)
2. Determinar el índice I que:
   - p1 + p2 + ... + p(I-1) < U <= p1 + p2 + ... + pI
3. Una vez seleccionado I, genere X usando la distribución Fi. 
4. retornar x

Si las probabilidades son del 20%, 30% y 50% respectivamente, la línea de tiempo se divide de la siguiente manera:

```text
0       0.20     0.50                             1.00

|--------|--------|--------------------------------|
    F1       F2                   F3
   20%      30%                  50%
```

Si el número fuera 0.10 cae en F1, si fuera 0.37 cae en F2 y si fuera 0.75 cae en F3.

## Algoritmo

```text
Generar U ~ Uniforme(0, 1)
acumulada = 0

Para i = 1 hasta n:
    acumulada = acumulada + p_i
    
    si U <= acumulada entonces
        generar X usando Fi
        retornar X
```

### (b)

En este caso podemos identificar que la función está formada por dos distribuciones diferentes:

$$
F(x)=
\begin{cases}
\frac{1-e^{-2x}+2x}{3}, & 0<x<1 \\
\frac{3-e^{-2x}}{3}, & x>1
\end{cases}
$$

Para $0<x<1$ podemos separarla así:

$$
F(x)=\frac{1}{3}(1-e^{-2x})+\frac{2}{3}x
$$

Entonces podemos identificar:

$$
F_1(x)=1-e^{-2x}
$$

que corresponde a una distribución exponencial con tasa:

$$
\lambda=2
$$

y también:

$$
F_2(x)=x
$$

que corresponde a una distribución uniforme entre 0 y 1.

Por lo tanto usamos:

$$
p_1=\frac{1}{3}
$$

$$
p_2=\frac{2}{3}
$$

Primero generamos $U_1 \sim U(0,1)$ para escoger cuál distribución utilizar.

Entonces:

- si $U_1 \leq \frac{1}{3}$, usamos $F_1$;
- si $U_1 > \frac{1}{3}$, usamos $F_2$.

Para generar la exponencial de tasa 2 usamos transformación inversa:

$$
X=-\frac{1}{2}\ln(U_2)
$$

y para la uniforme simplemente:

$$
X=U_2
$$

## algoritmo

```text
Generar U1 ~ Uniforme(0, 1)
Generar U2 ~ Uniforme(0, 1)

Si U1 <= 1/3 entonces
    X = -(1/2) ln(U2)

Si U1 > 1/3 entonces
    X = U2

retornar X
```


### (c)

En este caso tenemos una forma más general:

$$
F(x)=\sum_{i=1}^{n}\alpha_i x^i
$$

donde:

$$
\alpha_i \geq 0
$$

y:

$$
\sum_{i=1}^{n}\alpha_i=1
$$

Podemos considerar cada término como una distribución diferente:

$$
F_i(x)=x^i
$$

y cada distribución se escoge con probabilidad:

$$
p_i=\alpha_i
$$

Primero escogemos un índice $i$ utilizando las probabilidades:

$$
\alpha_1,\alpha_2,\ldots,\alpha_n
$$

Después generamos:

$$
U \sim U(0,1)
$$

Como:

$$
F_i(x)=x^i
$$

usamos transformación inversa:

$$
U=X^i
$$

y despejando $X$:

$$
X=U^{1/i}
$$

De esta forma podemos generar una variable aleatoria utilizando el término seleccionado.

## algoritmo

```text
Escoger un índice i usando las probabilidades
α1, α2, ..., αn

Generar U ~ Uniforme(0, 1)

X = U^(1/i)

retornar X
