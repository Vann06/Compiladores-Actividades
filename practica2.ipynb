# Ejercicio1 
Sea X una variable aleatoria exponencial con media 1. Proporcione un algoritmo eficiente para
simular una variable aleatoria cuya distribución es la distribución condicional de X dado que X <
0,05. Es decir, su función de densidad es 

Genere 1000 de estas variables y utilícelas para estimar E[X | X < 0,05]. Luego determine el
valor exacto de E[X | X < 0,05]

## Análisis:
Solo nos importan valores dentro del rango 0 < X < 0.05, que cumplen la condición. Debemos de encontrar la función acumulada:
 
 $$F(x) = \frac{1 - e^{-x}}{1 - e^{-0.05}}$$
para $0 < x < 0.05$

Ahora generamos una variable aleatoria uniforme $U$ en el intervalo $(0,1)$ y la transformamos usando la función inversa de $F(x)$:

$$U \sim \text{Uniforme}(0,1)$$

Para encontrar $X = F^{-1}(U)$, despejamos $x$ de la ecuación $U = F(x)$:

$$U = \frac{1 - e^{-x}}{1 - e^{-0.05}}$$

$$U(1 - e^{-0.05}) = 1 - e^{-x}$$

$$e^{-x} = 1 - U(1 - e^{-0.05})$$

$$-x = \ln[1 - U(1 - e^{-0.05})]$$

$$X = -\ln[1 - U(1 - e^{-0.05})]$$

## Explicación del algoritmo: 
1. Generar U uniforme en (0,1)
2. Calcular X usando la fórmula $X = -\ln[1 - U(1 - e^{-0.05})]$
3. Retornar X como la variable aleatoria simulada.

y ahora solo se generan 1000 de estas variables para estimar $E[X | X < 0.05]$ y luego determinar el valor exacto de $E[X | X < 0.05]$.

    

````python
n = 1000
suma = 0

# el valor esperado de X dado que X < 0.05 se puede estimar 
# generando n variables aleatorias uniformes U en el
# intervalo (0,1) y transformándolas usando la función inversa de F(x):
for i in range(n):
    u = random.random()

    x = -math.log(1 - u * (1 - math.exp(-0.05)))

    suma += x

promedio = suma / n

print("Estimación de E[X | X < 0.05]:", promedio)

# valor exacto de E[X | X < 0.05] se puede calcular usando la integral de la función de densidad condicional:
exacto = (1 - 1.05 * math.exp(-0.05)) / (1 - math.exp(-0.05))
print("Valor exacto de E[X | X < 0.05]:", exacto)
````

Generamos una distribución uniforme U ~ Uniforme(0,1) y contruimos intervalos acumulados correpondiendo a cada una de las distribuciones Fi, i = 1, . . . , n. un algoritmo para esto sería: 
1. Generar U uniforme en (0,1)
2. Determinar el índice I que:
   - p1 + p2 + ... + p(I-1) < U <= p1 + p2 + ... + pI
3. Una vez seleccionado I, genere X usando la distribución Fi.

# Ejercicio 4 
Una compañía de seguros contra siniestros tiene 1000 asegurados, cada uno de los cuales presentará una reclamación en el próximo mes de manera independiente con probabilidad 0.05. 

Suponiendo que los montos de las reclamaciones son variables aleatorias exponenciales independientes con media $800, utilice simulación para estimar la probabilidad de que la suma de estas reclamaciones exceda $50,000

## análisis: 

En este ejercicio tenemos una compañía con **1000 asegurados**.

Cada asegurado puede presentar una reclamación con probabilidad:

$$
p = 0.05
$$

Esto significa que para cada asegurado podemos generar un número aleatorio:

$$
U \sim U(0,1)
$$

y si:

$$
U < 0.05
$$

entonces consideramos que esa persona sí presenta una reclamación.

Si presenta una reclamación, el monto sigue una distribución exponencial con media:

$$
800
$$

Para una distribución exponencial:

$$
E[X] = \frac{1}{\lambda}
$$

por lo tanto:

$$
\lambda = \frac{1}{800}
$$

Para generar el monto de una reclamación podemos usar transformación inversa:

$$
X = -800\ln(U)
$$

donde $U$ es otro número uniforme entre 0 y 1.

La idea de la simulación es recorrer los 1000 asegurados, verificar cuáles presentan reclamación y sumar todos los montos generados.

Al terminar tenemos el total de reclamaciones de un posible mes.

Después verificamos si:

$$
\text{total} > 50000
$$

Pero como queremos estimar una probabilidad, no basta con hacer una sola simulación.  
Debemos repetir todo el proceso muchas veces, por ejemplo 10000 veces.

En cada simulación contamos si el total superó los \$50,000.

Finalmente estimamos la probabilidad con:

$$
P(\text{total} > 50000)
\approx
\frac{\text{cantidad de simulaciones donde el total superó 50000}}
{\text{cantidad total de simulaciones}}
$$

## Algoritmo

```text
repetir muchas simulaciones

    total = 0

    para cada uno de los 1000 asegurados

        generar U ~ Uniforme(0,1)

        si U < 0.05 entonces

            generar otro U ~ Uniforme(0,1)

            monto = -800 ln(U)

            sumar monto al total

    si total > 50000 entonces
        aumentar contador

probabilidad = contador / número de simulaciones
```

En promedio se esperan:

$$
1000(0.05)=50
$$

reclamaciones por mes.

Como cada reclamación tiene una media de \$800, el total promedio esperado sería aproximadamente:

$$
50(800)=40000
$$

Por eso tiene sentido que el ejercicio pregunte por la probabilidad de superar los \$50,000, ya que es un valor mayor al total promedio esperado.

````python
# usar simulación para P(suma de reclamos > 50000)
simulaciones = 10000
exceden_50000 = 0

for i in range(simulaciones):
    total = 0 
    for aseguradora in range(1000):
        # random reclamo o no 
        u = random.random()
        if u < 0.05:
            # generar reclamo exponencial con media 800
            u_monto = random.random()
            monto = -800 * math.log(u_monto)
            total += monto
    # revisa si la suma excede 50000
    if total > 50000:
        exceden_50000 += 1

probabilidad = exceden_50000 / simulaciones
print("Estimación de P(suma de reclamos > 50000):", probabilidad)
````

Los montos se suman y se verifica si el total supera los \$50,000. Se repite el proceso muchas veces para estimar la probabilidad de que esto ocurra. Tuvimos como resultado que la probabilidad de que la suma de las reclamaciones exceda \$50,000 es aproximadamente 0.1055, lo que significa que hay un 10.55% de probabilidad de que esto ocurra en un mes dado.

# Ejercicio 5 
Escriba un programa que genere variables aleatorias normales utilizando el método del Ejemplo
5f (método de rechazo con una distribución exponencial de tasa 1).

En este ejercicio queremos generar una variable aleatoria normal estándar:

$$
Z \sim N(0,1)
$$

El método usa **aceptación y rechazo** tomando como distribución auxiliar una exponencial de tasa 1:

$$
Y \sim Exp(1)
$$

Primero generamos un valor exponencial:

$$
Y = -\ln(U_1)
$$

donde:

$$
U_1 \sim U(0,1)
$$

Luego generamos otro número uniforme:

$$
U_2 \sim U(0,1)
$$

El valor de $Y$ se acepta solamente si cumple:

$$
U_2 \leq e^{-\frac{(Y-1)^2}{2}}
$$

Si no cumple esta condición, se rechaza el valor y se vuelve a generar otro.

Una vez que tenemos un valor aceptado, todavía necesitamos decidir si la normal será positiva o negativa.

Para esto generamos otro número:

$$
U_3 \sim U(0,1)
$$

Entonces:

- si $U_3 \leq 0.5$, usamos:

$$
Z=Y
$$

- si $U_3 > 0.5$, usamos:

$$
Z=-Y
$$

De esta forma obtenemos una variable:

$$
Z \sim N(0,1)
$$

## Algoritmo

```text
repetir

    generar U1 ~ Uniforme(0,1)

    Y = -ln(U1)

    generar U2 ~ Uniforme(0,1)

hasta que:

    U2 <= exp(-(Y - 1)^2 / 2)


generar U3 ~ Uniforme(0,1)

si U3 <= 0.5 entonces
    Z = Y

si U3 > 0.5 entonces
    Z = -Y

retornar Z
```

La idea del método es generar primero valores positivos usando una distribución exponencial y luego aceptar solamente aquellos que cumplen la condición de rechazo.

Después se escoge aleatoriamente el signo positivo o negativo para obtener finalmente una variable normal estándar.

Si queremos comprobar que el método funciona, podemos generar muchas variables y verificar que su media sea aproximadamente:

$$
E[Z] \approx 0
$$

y su varianza aproximadamente:

$$
Var(Z) \approx 1
$$

````python
# definimos la función para generar variables aleatorias 
# normales usando el método de rechazo con una distribución 
# exponencial de tasa 1
def normal_rechazo():

    while True:
        # generamos random u1 y calculamos y = -ln(u1)
        u1 = random.random()
        y = -math.log(u1)

        u2 = random.random()
        
        if u2 <= math.exp(-((y - 1) ** 2) / 2):
            break

    u3 = random.random()
    # verificamos si u3 <= 0.5 para determinar el signo de z
    if u3 <= 0.5:
        z = y
    else:
        z = -y

    return z


print(normal_rechazo())
````

Con este resultado de normalidad podemos notar un resultado de 0.899 lo que significa que se generó una observacion con ese normal estándar. 
Esto nos demuestra que el método de rechazo con una distribución exponencial de tasa 1 es efectivo para generar variables aleatorias normales estándar.

# Ejercicio 6
Escriba un programa que genere las primeras T unidades de tiempo de un proceso de Poisson
con tasa λ

### Proceso de Poisson 
Eventos que ocurren aleatoriamente en un tiempo determinado.
Para este caso tenemos que la tada es de 2 eventos por unidad de tiempo. 
 X ~ Exp(λ)

 ```text
 t = 0 
 mientras t < T: 
    generar U ~ Uniforme(0,1)
    generar X = -ln(U)/λ
    t = t + X
    
    si t <=T:
        guardar t

 ```



````python
lam = 2 
T = 20 

t = 0 
eventos = []

while t < T:
    U = random.uniform(0, 1)
    X = -(1/lam) * math.log(U)
    t = t + X

    if t <= T:
        eventos.append(t)

print(eventos)
````

estos resultados no son los tiempos entre eventos, sino los momentos en los que ocurrió cada evento. Por ejemplo, si el primer evento ocurre en t=0.5 y el segundo en t=1.2, entonces los tiempos entre eventos serían 0.5 y 0.7, respectivamente.

Esto lo que significa es que al generar la exponencial con tasa 2, estamos generando los tiempos entre eventos, y al ir sumando esos tiempos obtenemos los momentos en los que ocurren los eventos.
